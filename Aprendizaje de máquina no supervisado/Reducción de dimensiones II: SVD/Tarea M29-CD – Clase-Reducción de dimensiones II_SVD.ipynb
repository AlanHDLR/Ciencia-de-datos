```python
import pandas as pd
from numpy import array
from sklearn.decomposition import TruncatedSVD
from numpy import diag
from numpy import zeros
from scipy.linalg import svd
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d
%matplotlib inline
from mpl_toolkits.mplot3d import Axes3D
```

```python
df = pd.DataFrame(
  {
    "Matemáticas": [7.0, 7.5, 7.6, 5.0, 6.0, 7.8, 6.3, 7.9, 6.0, 6.8],
    "Ciencias": [6.5, 9.4, 9.2, 6.5, 6.0, 9.6, 6.4, 9.7, 6.0, 7.2],
    "Español": [9.2, 7.3, 8.0, 6.5, 7.8, 7.7, 8.2, 7.5, 6.5, 8.7],
    "Historia": [8.6, 7.0, 8.0, 7.0, 8.9, 8.0, 9.0, 8.0, 5.5, 9.0],
    "EdFísica": [8.0, 7.0, 7.5, 9.0, 7.3, 6.5, 7.2, 6.0, 8.7, 7.0],
  },
  index=["Lucía", "Pedro", "Inés", "Luis", "Andrés", "Ana", "Carlos", "José", "Sonia", "María"]
)
df
```

```python
A = df
A
```

```python
 U, s, VT = svd(A) # Obtencion de matrices cuyo producto arroja la matriz A
Sigma = zeros((A.shape[0], A.shape[1])) # Creación de matriz de 0 de la misma dimension que A
Sigma[:A.shape[1], :A.shape[1]] = diag(s)
```

```python
print("Matriz U:")
print(U)
```

```python
print("Matriz Sigma")
print(Sigma)
```

```python
print("Matriz VT")
print(VT)
```

```python
# Para reconstruir la matriz originaL y verificar que todo bien
# .dot es la forma de multiplicar 
B = U.dot(Sigma.dot(VT))
print(B)
```

```python
# Reducción a dos simensiones
n_elements = 2
UReduced2 = U[:, :n_elements]
print(UReduced2)
```

```python
SigmaReduced2 = Sigma[:n_elements,:n_elements]
print("Matriz Sigma Reducida")
print(SigmaReduced2)
```

```python
VTReduced2 = VT[:n_elements,:]
print(VTReduced2)
```

```python
AReduced2 = UReduced2.dot(SigmaReduced2.dot(VTReduced2))
print("Maatriz A transformada: ")
print(AReduced2)
```

```python
T2 = UReduced2.dot(SigmaReduced2)
print("Matriz T (2 Dimensiones): ")
print(T2)
```

```python
df = pd.DataFrame(T2, columns = ["a", "b"])
df.plot(kind = "scatter", x = "a", y = "b")
plt.show()
```

```python
etiquetas = ["Lucía", "Pedro", "Inés", "Luis", "Andrés", "Ana", "Carlos", "José", "Sonia", "María"]
# Agregamos etiquetas
df = pd.DataFrame(T2, columns = ["a", "b"])
x = df.iloc[:,0]
y = df.iloc[:, 1]
x = x.to_numpy()
y = y.to_numpy()

fig, ax = plt.subplots()
ax.set_title("Mapa de observaciones")
ax.scatter(x,y)

for i, txt in enumerate(etiquetas):
    ax.annotate(txt, (x[i], y[i]))

plt.show()
```

```python
# Reducción a 3 dimensiones
n_elements = 3
UReduced3 = U[:, :n_elements]
#print("U reducida:")
#print(UReduced3)

SigmaReduced3 = Sigma[:n_elements,:n_elements]
#print("Matriz Sigma Reducida")
#print(SigmaReduced3)

VTReduced3 = VT[:n_elements,:]
#print("VT Reducida")
#print(VTReduced3)

AReduced3 = UReduced3.dot(SigmaReduced3.dot(VTReduced3))
print("Maatriz A transformada: ")
print(AReduced3)
```

```python
T3 = UReduced3.dot(SigmaReduced3)
print("Matriz T (3 Dimensiones): ")
print(T3)
```

```python
df3 = pd.DataFrame(T3, columns = ["a", "b", "c"])
df3
```

```python
# Creación de la grafica en 3d
fig = plt.figure()
ax = plt.axes(projection="3d")
xline = df3["b"]
yline = df3["c"]
zline = df3["a"]
ax.scatter3D(xline, yline, zline)
plt.show()
```

```python
etiquetas = ["Lucía", "Pedro", "Inés", "Luis", "Andrés", "Ana", "Carlos", "José", "Sonia", "María"]
from mpl_toolkits.mplot3d import Axes3D

ax3d = plt.figure().add_subplot(projection="3d")

arrayx = np.array(df3["a"])
arrayy = np.array(df3["b"])
arrayz = np.array(df3["c"])

ax3d.scatter(arrayx, arrayy, arrayz)

for x, y, z, label in zip(arrayx, arrayy, arrayz, etiquetas):
    if label == "Setosa":
        ax3d.text(x, y, z, label)

plt.title("Datos 3D")
plt.show(ax3d)
```
