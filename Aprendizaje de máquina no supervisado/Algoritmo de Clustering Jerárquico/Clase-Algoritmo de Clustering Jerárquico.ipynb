```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os 
```

```python
os.chdir("/Users/AlanHDLR/Desktop")
data = pd.read_csv("Wholesale customers data.csv")

data.head()
```

```python
# Vamos a normalizar los datos, para tenerlos todos en una misma escala
from sklearn.preprocessing import normalize
data_scaled = normalize(data)
data_scaled
```

```python
# Agregamos los titulos del DF original 
data_scaled = pd.DataFrame(data_scaled, columns = data.columns)
data_scaled.head()
```

```python
# Vamos a crear el cluster
import scipy.cluster.hierarchy as shc
plt.figure(figsize = (10,7))
plt.title("Dendograma")
dend = shc.dendrogram(shc.linkage(data_scaled, method='ward'))
```

```python
# Verificar cuentos grupos diferenets tenemos 
print(dend['color_list'])
# C0 siempre vaa  ser la unión de todos los grupos, (la ultima)
```

```python
colores_unicos = set(dend['color_list'])
colores_unicos
```

```python
# Para saber cual es el número de clusters optimos, siempre hay que contar el número de colores y restarle 1
num_clusters_optimo = len(colores_unicos) - 1
num_clusters_optimo
```

```python
plt.figure(figsize = (10,7))
plt.title("Dendograma")
dend = shc.dendrogram(shc.linkage(data_scaled, method = 'ward'))
plt.axhline(y = 6, color = 'r', linestyle = '--')
```

```python
data
```

```python
# Obtenemos que elementos pertenecen a cada grupo
from sklearn.cluster import AgglomerativeClustering
cluster = AgglomerativeClustering(n_clusters = 2, metric = 'euclidean', linkage = 'ward')
grupos = cluster.fit_predict(data_scaled)
grupos
```

```python
# Visualizamos como se veria la separación de los datos
plt.figure(figsize = (10,7))
plt.scatter(data_scaled['Milk'], data_scaled['Grocery'], c = cluster.labels_)
```

```python
# Separación de grupos mediante PCA
campos = data_scaled.values
print(campos)
```

```python
from sklearn import decomposition
pca = decomposition.PCA(n_components = 2)
pca.fit(campos)
campos = pca.transform(campos)
campos
```

```python
plt.figure(figsize = (10,7))
plt.scatter(campos[:,0], campos[:,1], c = cluster.labels_)
```

```python
dataframe = pd.DataFrame(grupos, columns = ['grupo'])
dataframe
```

```python
dataframe2 = pd.concat([data, dataframe], axis = 1, join = 'inner')
dataframe2
```

# Análisis del caso Estudiantes 

```python
data = pd.DataFrame(np.array([[7, 6.5, 9.2, 8.6, 8],
                              [7.5, 9.4, 7.3, 7, 7],
                              [7.6, 9.2, 8, 8, 7.5],
                              [5, 6.5, 6.5, 7, 9],
                              [6, 6, 7.8, 8.9, 7.3],
                              [7.8, 9.6, 7.7, 8, 6.5],
                              [6.3, 6.4, 8.2, 9, 7.2],
                              [7.9, 9.7, 7.5, 8, 6],
                              [6, 6, 6.5, 5.5, 8.7],
                              [6.8, 7.2, 8.7, 9, 7]]),
                    columns=['Matematicas', 'Ciencias', 'Español', 'Historia', 'EdFisica'])
data   
```

```python
# Normalizamos los datos 
from sklearn.preprocessing import normalize
data_scaled = normalize(data)
data_scaled
```

```python
# Vamos a crear el cluster
import scipy.cluster.hierarchy as shc
plt.figure(figsize = (10,7))
plt.title("Dendograma")
dend = shc.dendrogram(shc.linkage(data_scaled, method='ward'))
```

```python
colores_unicos = set(dend['color_list'])
num_clusters_optimo = len(colores_unicos) - 1
num_clusters_optimo
```

```python
plt.figure(figsize = (10,7))
plt.title("Dendograma")
dend = shc.dendrogram(shc.linkage(data_scaled, method = 'ward'))
plt.axhline(y = 0.2, color = 'r', linestyle = '--')
```

```python
# Anlisis grafico con PCA
campos = data_scaled
from sklearn import decomposition
pca = decomposition.PCA(n_components = 2)
pca.fit(campos)
campos = pca.transform(campos)
campos
```

```python
# Obtenemos que elementos pertenecen a cada grupo
from sklearn.cluster import AgglomerativeClustering
cluster = AgglomerativeClustering(n_clusters = 3, metric = 'euclidean', linkage = 'ward')
grupos = cluster.fit_predict(data_scaled)
grupos
```

```python
plt.figure(figsize = (10,7))
plt.scatter(campos[:,0], campos[:,1], c = cluster.labels_)
```

```python
dataframe = pd.DataFrame(grupos, columns = ['grupo'])
dataframe
```

```python
dataframe2 = pd.concat([data, dataframe], axis = 1, join = 'inner')
dataframe2
```
