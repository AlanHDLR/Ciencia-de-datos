```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
import os 

os.chdir("/Users/AlanHDLR/Desktop")

from sklearn.cluster import KMeans

dataset = pd.read_csv("Mall_Customers.csv")
dataset.head()
```

```python
fields = dataset.iloc[:,[3,4]].values
print(fields)
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
# Calculara la inercia que hay, tomando en cuenta los puntos que hay
wcss = []
for index in range(1,11):
    kmeans = KMeans(n_clusters = index, init = "k-means++", random_state = 2)
    kmeans.fit(fields)

    wcss.append(kmeans.inertia_)
```

```python
# Graficación del "Codo de Jambu"
sb.set()
plt.plot(range(1,11), wcss)
plt.title("K-Means Clustering")
plt.xlabel("Número de Clusters")
plt.ylabel("WCSS") # Weighted cluster sum of squares
```

```python
kmeans = KMeans(n_clusters= 5, init = "k-means++", random_state = 42)
cluster_values = kmeans.fit_predict(fields)
print(cluster_values)
```

```python
plt.figure(figsize = (10,10))
plt.scatter(fields[cluster_values == 0,0], fields[cluster_values == 0,1], color = "blue", label ='Cluster 1')
plt.scatter(fields[cluster_values == 1,0], fields[cluster_values == 1,1], color = "orange", label ='Cluster 2')
plt.scatter(fields[cluster_values == 2,0], fields[cluster_values == 2,1], color = "green", label ='Cluster 3')
plt.scatter(fields[cluster_values == 3,0], fields[cluster_values == 3,1], color = "yellow", label ='Cluster 4')
plt.scatter(fields[cluster_values == 4,0], fields[cluster_values == 4,1], color = "red", label ='Cluster 5')

# Centroides
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1], color = 'black', label = 'Centroides')

plt.title("Grupos de clientes")
plt.xlabel = "Annuel income"
plt.ylabel = "Spending Score"
plt.legend()
```

```python
# Identificar el número de grupos optimo 
# Criterio de Silueta ("Silhouette")
from sklearn import datasets
from sklearn.metrics import silhouette_score
```

```python
X = dataset.iloc[:, [3,4]].to_numpy()
for j in range (2,12):
    kmeans = KMeans(n_clusters = j, random_state = 42)
    kmeans.fit_predict(X)
    # Calculamso el score de Silueta
    score = silhouette_score(X, kmeans.labels_, metric = 'euclidean')
    print("Score Silhouette;", "k = ", j, ":", score)
```

```python
# Como ultilizar más de 2 columnas de datos
# Utilizar PCA para convertir a 2 variables
# Algortimo KMeans aplicado para 3 columnas bajo una treasnformación de PCA

fields2 = dataset.iloc[:, [2,3,4]].values
print(fields2)
```

```python
# Utilización de PCA para convertir de 2 columnas a 3
from sklearn import decomposition

pca = decomposition.PCA(n_components = 2)
pca.fit(fields2)
fields2 = pca.transform(fields2)
fields2
```
