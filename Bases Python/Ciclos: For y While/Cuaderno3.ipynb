```python
x = float(input("Dame un número: "))

if x < 2:
    print("x = ", x)
    print("Pequeño")
# elif nos sirve para checvra mas de 3 condiciones
elif x < 10:
    print("x = ", x)
    print("Mediano")
else:
    print("x = ", x)
    print("Grande")
# \n es equivalente a un salto de línea
print("\nfin")
```

```python
# Como hacer el codigo mas sencillo y compacto
x = float(input("Dame un número: "))
print("x = ", x)
if x < 2: print("Pequeño")
elif x < 10: print("Mediano")
else: print("x = ", x)
print("\nfin")
```

```python
#Se pueden usar tantos elif como se desean
x = float(input("Dame un número: "))
print("x = ", x)
if x < 2: print("Pequeño")
elif x < 10: print("Mediano")
elif x < 20: print("Grande")
else: print("Gigante")
print("\nfin")
```
