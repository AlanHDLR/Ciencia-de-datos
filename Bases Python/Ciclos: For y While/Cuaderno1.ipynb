# Mamejo de disyunciones (OR)
## Con que una de las expresiones sea verdera, toda la instruccion sera verdera

```python
a = -5
b = 4
```

```python
# Verdadera y verdadera
a < 0 or b > 0
```

```python
# Falsa y verdadera
a > 0 or b > 0
```

```python
# Verdadera y verdadera
a < 0 or b < 0
```

```python
# Falso y falso
a > 0 or b < 0
```

```python
import numpy as np

a = np.array([True, False, True])

b = np.array([False, False, True])

# Uso de & y | para operaciones lógicas en arrays booleanos

result_and = a & b # Resultado: array([False, False, True])

result_or = a | b # Resultado: array([ True, False, True])

print(result_and)

print(result_or)
```
