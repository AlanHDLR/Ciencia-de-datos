```python
x = 5 
if x == 5:
    print("Igual a 5")
if x > 4:
    print("Mayor que 4")
if x >= 5:
    print("Mayor o igual a 5")
if x < 6:
    print("Menor a 6")
if x <= 5:
    print("Menor o igual a 5")
if x != 6:
    print("Diferente de 6")
```
