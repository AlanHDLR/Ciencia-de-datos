# Estructura TRY y EXCEPT
## Se utilizan para que cuando exista un error, haya una puerta de salida

```python
cadena1 = "123"
entero = int(cadena1)
print(entero)
```

```python
cadena = input("Dame un valor entero positivo: ")
try: 
    entero = int(cadena)
except:
    entero = -1

if entero > 0:
    print("Valor correcto")
else:
    print("Valor incorrecto")
```

# Usos avanzados de IF

```python
lluvia = False
print("Vamos a", "la playa" if not lluvia else "la biblioteca")
```

```python
lluvia = True
print("Vamos a", "la playa" if not lluvia else "la biblioteca")
```

```python
x = y = 40
z = x + 1 if x > y else y + 2
z
```

```python
# Como realizar instancias multiples
x = 3
s = ("abc" if (x == 1) else
     "def" if (x == 2) else
     "ghi" if (x == 4) else
     "mno")
s
```
