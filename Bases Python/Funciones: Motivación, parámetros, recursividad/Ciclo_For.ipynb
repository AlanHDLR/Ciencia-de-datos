```python
# A esto se el cocnoce como programacion orienta a objetos 
# En donde el objeto es el valor mas pequeñoq eu tiene una lista (valor)
# En esta caso, la variable "valor" va a tomar cada elemento de la lista y los va a gregando a la instrucción
numeros = [6, 5, 3, 8, 4, 2, 5, 4, 11]
suma = 0
for valor in numeros:
    suma = suma + valor
print("La suma es: ", suma)
    
```

```python
# La variable "fruta" puede ser cualquier nombre "x", "y" 
# Al valor mas pequeño de cada lista se le conoce como "átomo"
frutas = ["manzana", "fresa", "cereza"]
for fruta in frutas:
    print(fruta)
```

```python
total = 0
for i in 5, 7, 11, 13:
    print(i)
    total += i
print("Total: ", total)
```

```python
# Al poner "in range" no se tomara en cuenta el limite superior
# Se le peude poner solo un limite superior
for x in range(6):
    print(x)
```

```python
# O se le puede especificar un rango como tal de valores
for x in range(2, 10):
    print(x)
```

```python
# Podemos modificar cada cuanto se incrementan los valores siendo el ultimo intervalor el valor en el cual se incrementan estos númeoros
for x in range(2, 10, 3):
    print(x)
```

```python
# Podemos convinar con un "else" para que una vez que termine el ciclo realice una instrucción
for x in range(6):
    print(x)
else:
    print("Proceso terminado")
```

```python
# Ejercicio
# Deteremina rmediente un ciclo "for" el número más grande en la lista siguiente:
# 9, 41, 12, 3, 74, 15

mas_grande = -1
for numero in [9, 41, 12, 3, 74, 15]:
    if numero > mas_grande:
        mas_grande = numero
print("El número mas grande es: ", mas_grande)
```

```python
# Podemos hacer que un ciclo se evalue hasta que llegue a un resultado en este caso "fresa"
frutas = ["manzana", "fresa", "cereza"]
for fruta in frutas:
    print(fruta)
    if fruta == "fresa":
        break
```

```python
for x in range(6):
    if x == 3: break
    print(x)
else:
    print("Proceso terminado")
```

```python
nombre = input("Nombre del estudiante: ")
# Creación de un diccionario
datos = {"Jose" : 85, "Angela" : 92, "Miguel" : 75, "Ana" : 98}

for estudiante in datos:
    if estudiante == nombre:
        print("Calificación: ", datos[estudiante])
        break
else:
    print("No se encontro al estrudiante")
```

```python
# Ciclos anidados 
# Esta es una manera de tener un for dentro de otro (anidados)
# Se le asiganara cada adjetivo a cada fruta
frutas = ["manzana", "fresa", "cereza"]
adjetivos = ["roja", "grande", "deliciosa"]

for fruta in frutas:
    for tipo in adjetivos:
        print(fruta, tipo)
```

```python
# La función "reversed" nos da la posibilidads de invertir una lista
lista = [10, 20, 30, 40]
for numero in reversed(lista):
    print(numero)
```

```python
tablas = int(input("Número de tablas a calcular:" ))

for i in range(1, tablas + 1):
    print("Tabla de multiplicar del ",i ,":")
    contador = 1
    while contador < 11:
        print(i * contador, end = ", ")
        contador += 1 
    print("\n")
```
