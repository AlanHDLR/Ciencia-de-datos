# While
## Los ciclos o bucles "while" se les llama ciclos indefinidos por que se mantienen en ejecución hasta que una condición se vuelva falsa
### Siempre se tiene que prevenir que no se entre en un buvle infinito con una instrucción

```python
n = 4
while n > 0:
    print(n)
    n = n -1
print("Fin")
```

```python
# Aqui podemos ver una condicion eterna ya que el valor de n nunca cambia por lo que la isntruccion se ejecutara eternamente 
n = 4
while n > 1:
    print("Uno")
    print("Dos")
print("Fin")
```
