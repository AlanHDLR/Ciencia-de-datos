```python
n = int(input("Captura el valor de n: "))

suma = 0
i = 1

while i <= n:
    suma += i
    i += 1
print("La suma es: ", suma)
```

```python
numeros = (1,2,3,4,5,6,7,8,9)
valores = len(numeros)
impares = 0
pares = 0

contador = 1
while contador <= valores:
    if numeros[contador - 1] % 2 == 0:
        pares += 1
    else:
        impares += 1
    contador += 1
    
print("La cantidad de numeros pares es: ", pares)
print("La cantidad de numero simpares es: ", impares)
```
