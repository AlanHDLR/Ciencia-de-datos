```python
# Concatenar dos valores
nombre = "Jose "
nombre2 = "Raúl"
nombre3 =  nombre + nombre2
print(nombre3)
```

```python
# Slicing o extracción de caracteres
print(nombre3[1])
```

```python
print(nombre3[0])
```

```python
# El limite inferiro si se toma en cuenta 
# Longitud de la cadena = 8 valores
print(nombre3[0:9])
```

```python
# El limite superior no se toma en cuenta, por lo que se tiene que poner 1 valor más
print(nombre3[0:8])
```

```python
print(nombre3[0:3])
```

```python
print(nombre3[0:4])
```

```python
# Si no le ponemos limite superior, nos imprimiratoda la cadena posterior al valor de referencia de inicio
print(nombre3[0:])
```

```python
# Si no se coloca ningun valor, imprimira por consecunacia toda la cadena 
print(nombre3[:])
```

```python
# NO podemos pedir mas valores de los que tenga en total la cadena 
print(nombre3[20])
```

```python
# Para poder conocer la longitud de una cadena, utilizamos la funcion "len()"
print(len(nombre3))
```

```python
# ".lower()" nos sirve para volver todas las letras a minusculas 
print(nombre3.lower())
```

```python
# ".lower()" nos sirve para volver todas las letras a mayusculas
print(nombre3.upper())
```

```python
# Si no sabemos que comandos se pueden aplicar a una variable, podemos usar "dir()"
dir(nombre3)
```

# Uso de metodos 

```python
# capitalize() : Convierte el primer carácter de la cadena a mayúscula.

texto = "hola amigo"
print(texto.capitalize()) # Salida: "Hola amigo"
```

```python
# count() : Cuenta el número de ocurrencias de una subcadena.

texto = "Hola amigo"
print(texto.count('a')) # Salida: 2
```

```python
# find() : Encuentra la posición de una subcadena.

texto = "Hola amigo"
print(texto.find('amigo')) # Salida: 5
```

```python
# replace() : Sustituye una subcadena por otra.

texto = "Hola amigo"
texto_modificado = texto.replace("amigo", "amiga")
print(texto_modificado) # Salida: "Hola amiga"
```
