# Tuplas

* Una tupla es un tipo de secuencia que trabaja de manera similar a una lista pero en la cual no se puede alterar su contenido (de manera similar a una cadena). Usamos paréntesis para definirla. 
* Por dicha razón se dice que las tuplas son “inmutables” (inmodificables). 
* Debido que las tuplas no se pueden modificar, son más eficientes en términos de memoria y rendimiento que las listas (especialmente al manejar variables temporales).

```python
diccionario = dict()
diccionario["Alberto"] = 25
diccionario["Pedro"] = 30
diccionario
```

```python
#Convertir diccionarios en tuplas
tuplas = diccionario.items()
tuplas
```

```python
# Ordenamiento de tuplas, se les asigan un valor con ":" 
d = {"d" : 10, "b" : 15, "g" : 25}
d.items()
```

```python
tuplas = sorted(d.items())
tuplas
```
