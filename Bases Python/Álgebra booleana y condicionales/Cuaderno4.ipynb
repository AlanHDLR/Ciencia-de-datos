# Listas 

* Coleccion de objetos de diversa indole
* Losobjetospuedensernúmeros, letras e incluso otras listas. 
* Laslistasson“mutables” en el sentido de que sus elementos pueden ser intercambiados en cualquier momento. 
* Esposibleaplicar operaciones de ordenamiento a sus elementos.

```python
amigos = ["Pedro", "Juan", "Jose"]
amigos 
```

```python
# Se peuden mezclar variso tipos de valores en las listas 
elementos = ["rojo", 15, [2,9]]
elementos
```

```python
# Podemos acceder a los valores de las listas con sus indices
print(amigos[1])
```

```python
# Al ser mutable, podemos modificar los elementos 
amigos_cercanos = amigos 
amigos_cercanos[2] ="Jesus"
amigos_cercanos
```

```python
# Podemos utilizar operadores con las listas y no es necesario, llamar un print, podemos guardarlo en una variable 
longitud = len(amigos)
longitud
```

```python
# Igualmente se pueden concatenar varias listas
a = [1,2,3]
b = [4,5,6]
c = a + b
c
```

```python
# Podemos rebanar listas 
c[1:3]
```

```python
c[2:]
```

```python
# Podemos declara una lista sin elementos dentro 
listado = list()
listado
```

```python
# Podemos agregar elementos posteriormente con ".append()"
listado.append("libro")
listado.append(100)
listado
```
