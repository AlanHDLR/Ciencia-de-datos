# Diccionario

* Un diccionario contiene una colección de elementos donde cada uno posee una etiqueta. 
* Los diccionarios son similares a las listas, con la diferencia de que utilizan llaves en vez de números para localizar valores. 
* Al igual que en las listas, es posible crear un diccionario desde cero e irlo construyendo gradualmente.


```python
elementos = dict()
elementos
```

```python
# Es como establecer una serie de codigos en deonde la primer parte es una llave y la segunda su valor 
elementos["dinero"] = 10000
elementos["dulces"] = 20
elementos["amigos"] = 5
elementos
```

```python
elementos["dulces"]
```

```python
#Principales diferencias entre listas y diccionarios
# Listas
lista = list()
lista.append(20)
lista
```

```python
# Para crear un diccionario se utiliza "dict()" o solo se colocan las llaves "{}"
diccionario = {}
diccionario["edad"] = 22
diccionario
```

```python
# Se puede utilizar esto para hacer analisis en textos, viendo que palabras se pueden repetir mas y en que partes
conteo = dict()
nombres = ["Juan", "Pedro", "Juan", "Ramiro"]
# El seudocodigo son los pasos que seguiremos para llegar a una solución 
# Para cada nombre en la lista de nombres
# Checar si el nombre ya existe en el diccionario "conteo"
# Si el nombre no esta en la lista, inicialiozar su frecuencia en 1 
# En caso contrario (ya esta en la lista), incrementar su frecuencia en 1 

for nombre in nombres:
    if nombre not in conteo:
        conteo[nombre] = 1
    else:
        conteo[nombre] = conteo[nombre] + 1
conteo
```

```python
# Abrir un archivo en Python
archivo = open("Trabalenguas.txt")
archivo
```

```python
# Asignar el archivo a una variable 
for texto in archivo:
    palabras = texto.split()
palabras
```

```python
conteo = dict()
for nombre in palabras:
    if nombre not in conteo:
        conteo[nombre] = 1
    else:
        conteo[nombre] = conteo[nombre] + 1
conteo
```
