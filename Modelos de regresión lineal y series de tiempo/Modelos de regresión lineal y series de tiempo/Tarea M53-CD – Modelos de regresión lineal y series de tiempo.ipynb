```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings("ignore")

import os
os.chdir("/Users/AlanHDLR/Desktop")
```

```python
data_vw = pd.read_csv("Advertising.csv")
data_vw
```

```python
data_vw["Intercepto"] = 1
X = data_vw[['Intercepto', 'TV', 'Radio', 'Newspaper',]].values
Y = data_vw[["Sales"]].values 
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.30, random_state=1)

print(f"Registros para entrenamiento: {len(X_train)}")
print(f"Registros para prueba: {len(X_test)}")
```

```python
import statsmodels.api as sm
modelo_completo = sm.OLS(y_train, X_train).fit()
print(modelo_completo.summary())
```

```python
# Nuevo conjunto X solo con variables significativas
X_opt = data_vw[['Intercepto', 'TV', 'Radio']].values 

# Volver a dividir (manteniendo random_state=1)
X_train_opt, X_test_opt, y_train, y_test = train_test_split(X_opt, Y, test_size=0.30, random_state=1)

# Ajustar el nuevo modelo
modelo_final = sm.OLS(y_train, X_train_opt).fit()
print(modelo_final.summary())
```

```python
XT_X = np.matmul(np.matrix.transpose(X_train_opt), X_train_opt)
XT_X
```

```python
XT_X_inv = np.linalg.inv(XT_X)
XT_X_inv
```

```python
XT_Y = np.matmul(np.matrix.transpose(X_train_opt), y_train)
XT_Y
```

```python
betas = np.matmul(XT_X_inv, XT_Y)
betas
```

```python
# Calculo de Y's pronosticadas
Y_pred = np.matmul(X_train_opt, betas)
Y_pred
```

```python
# Calculo de residuales
Resid = y_train - Y_pred
Resid.round(2)
```

```python
# Calculo de la suma de residuales al cuadrado
RSS = np.matmul(np.matrix.transpose(Resid), Resid)
RSS
```

```python
# Calculo de la suma total de cuadrados
TSS = np.matmul(np.matrix.transpose(y_train), y_train) - len(y_train)*(y_train.mean()**2)
TSS
```

```python
# Calculo de coeficiente de determinación 
R_cuad = float(1- RSS/TSS)
R_cuad
```

```python
# Calculo de la varianza del error de regresión
s_cuad = RSS / (len(y_train) - X_train_opt.shape[1])
s_cuad
```

```python
# Desviación estandar del error de regresión
import math
s = math.sqrt(s_cuad)
s
```

```python
import scipy.stats

# IMPORTANTE: Usa los datos de entrenamiento (70%) para los grados de libertad
# n = filas de entrenamiento, p = columnas de X (incluyendo intercepto)
n = len(y_train)
p = X_train_opt.shape[1] 
grados_libertad = n - p

# Para un intervalo del 90%, el nivel de confianza es 0.90
Confianza = 0.90 
q = (1 - Confianza) / 2 # Esto dará 0.05 para cada cola

# t_critico para el 90%
t_critico = abs(scipy.stats.t.ppf(q, df = grados_libertad))

print(f"Grados de libertad: {grados_libertad}")
print(f"Valor t crítico (90%): {t_critico:.4f}")
```

```python
# Vector de valores particulares de X
# Al eliminar la variable de "newspapper" ya que no era significativa, 
# no ponemos su valor en el ejemplo
f = np.array([[1], [100], [50]])
f
```

```python
Margen_error = t_critico * (s * (float(np.matmul(np.matmul(np.matrix.transpose(f), XT_X_inv), f)) ** 0.5))
Margen_error
```

```python
Pron_puntual = float(np.matmul(np.matrix.transpose(f), betas))
Pron_puntual
```

```python
# Limites del intervalo de confianza
Lim_inferior = Pron_puntual - Margen_error
Lim_superior = Pron_puntual + Margen_error
print("Intervalo de confienza: (", Lim_inferior, ",", Lim_superior, ")")
```

# Funciones para el calculo de pruebas

```python
# Sesgo y Curtosis
def mis_momentos(residuos):
    n = len(residuos)
    media_res = np.mean(residuos)
    
    # Numeradores (momentos 2, 3 y 4)
    m2 = np.sum((residuos - media_res)**2) / n
    m3 = np.sum((residuos - media_res)**3) / n
    m4 = np.sum((residuos - media_res)**4) / n
    
    sesgo = m3 / (m2**1.5)
    curtosis = m4 / (m2**2)
    
    return sesgo, curtosis

# Comparación
s, c = mis_momentos(modelo_final.resid)
print(f"Mi Sesgo: {s:.3f} | Mi Curtosis: {c:.3f}")
```

```python
# Durbin Watson
def mi_durbin_watson(residuos):
    numerador = np.sum(np.diff(residuos)**2)
    denominador = np.sum(residuos**2)
    return numerador / denominador

print(f"Mi DW: {mi_durbin_watson(modelo_final.resid):.3f}")
```

```python
from scipy import stats
def mi_prueba_white(X, residuos):
    n = len(residuos)
    e2 = residuos**2
    
    #  Crear matriz con términos cuadráticos e interacciones
    # Supongamos X tiene [Const, TV, Radio]
    tv = X[:, 1]
    radio = X[:, 2]
    
    # Matriz de diseño para White: [Const, TV, Radio, TV^2, Radio^2, TV*Radio]
    X_white = np.column_stack([X, tv**2, radio**2, tv*radio])
    
    # Regresión auxiliar
    modelo_aux = sm.OLS(e2, X_white).fit()
    
    # Estadístico LM = n * R^2
    lm_stat = n * modelo_aux.rsquared
    p_val = 1 - stats.chi2.cdf(lm_stat, df=X_white.shape[1] - 1)
    
    return lm_stat, p_val

# Uso (X_train_final incluye la columna de 1s)
lm, p = mi_prueba_white(X_train_opt, modelo_final.resid)
print(f"White LM: {lm:.4f}, p-value: {p:.4f}")
```
