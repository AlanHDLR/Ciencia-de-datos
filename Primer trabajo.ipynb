## Segunda parte
### Realizar la limpieza de datos

```python
# DataFrame "DIM_CATEGORY"

# Todas las tablas tienen los nombres de sus columnas bien escritas por lo que los errores tipograficos no existen
# En caso de que si, podemos renombrarlas
# Para este DataFrame en cuestión solo cambiaremos los nombres a español
DIM_CATEGORY.columns = ['# CATEGORIA', 'CATEGORIA']

# No existen valores nulos ya que la tabla es muy pequeña y se puede visualizar a simple vista 

# No existen duplicados ya que se puede ver a simple vista todos los datos de la tabla 

# El unico cambio a realizar en esta tabla sería modificar el nombre de las categorias para pasarlas a español
DIM_CATEGORY['CATEGORIA'] = DIM_CATEGORY['CATEGORIA'].replace('FABRIC TREATMENT AND SANITIZER', 'TRATAMIENTO Y DESINFECTANTE DE TELAS')
# Creo que la segunda categoria no la dectecta por que el  "I" no es igual al valor del DataFrame 
DIM_CATEGORY['CATEGORIA'] = DIM_CATEGORY['CATEGORIA'].replace('AIR CARE', 'CUIDADO DEL AIRE')
DIM_CATEGORY['CATEGORIA'] = DIM_CATEGORY['CATEGORIA'].replace('LAVATORY CARE & BRC', 'CUIDADO DE LAVADORES Y BRC')
DIM_CATEGORY
DIM_CATEGORY.to_csv('DIM_CATEGORY.csv', index=False)
```

```python
# DataFrame "DIM_PRODUCT"

# Todas las tablas tienen los nombres de sus columnas bien escritas por lo que los errores tipograficos no existen
# En caso de que si, podemos renombrarlas
# Para este DataFrame en cuestión solo cambiaremos los nombres a español
DIM_PRODUCT.columns = ["FABRICANTE", "MARCA", "ARTICULO", "DESCRIPCION_ARTICULO", "CATEGORIA", "PRESENTACION", "ATRIBUTO_1", "ATRIBUTO_2", "ATRIBUTO_3"]

# Verificamos si existen valores nulos con ".isnull().sum()" y de ser así, se rellenaran con "NO VALIDO" 
DIM_PRODUCT.isnull().sum()
DIM_PRODUCT.fillna("NO VALIDO", inplace=True)

# Verificamos si existen datos duplicados
# Eliminar duplicados con la función ".drop_duplicates"
# Al incorporar "inplace=True" aplica los datos directamente en el DataFrame original 
DIM_PRODUCT.drop_duplicates(inplace = True)

# Por ultimo cambiamos los datos dentro del DataFrame de Ingles a Español
DIM_PRODUCT = DIM_PRODUCT.replace({"FABRIC TREATMENT": "TRATAMIENTO DEL TEJIDO", "SAFE BLEACH": "LEJÍA SEGURA",
                                    "GEL": "LÍQUIDO Y GEL", "PRELAVADOR": "PRETRATAMIENTO"})
DIM_PRODUCT
DIM_PRODUCT.to_csv('DIM_PRODUCT.csv', index=False)
```

```python
# DataFrame "DIM_SEGMENT"

# Todas las tablas tienen los nombres de sus columnas bien escritas por lo que los errores tipograficos no existen
# En caso de que si, podemos renombrarlas
# Para este DataFrame en cuestión solo cambiaremos los nombres a español
DIM_SEGMENT.columns = ["CATEGORIA", "ATRIBUTO_1", "ATRIBUTO_2", "ATRIBUTO_3", "PRESENTACION", "SEGMENTO"]

# Verificamos si existen valores nulos con ".isnull().sum()" y de ser así, se rellenaran con "NO VALIDO" 
DIM_SEGMENT.isnull().sum()
DIM_SEGMENT.fillna("NO VALIDO", inplace=True)

# Verificamos si existen datos duplicados
# Eliminar duplicados con la función ".drop_duplicates"
# Al incorporar "inplace=True" aplica los datos directamente en el DataFrame original 
DIM_SEGMENT.drop_duplicates(inplace = True)

# Por ultimo cambiamos los datos dentro del DataFrame de Ingles a Español
DIM_SEGMENT = DIM_SEGMENT.replace({"BLEACH": "LEJÍA", "FABRIC TREATMENT": "TRATAMIENTO DEL TEJIDO", "SAFE BLEACH": "LEJÍA SEGURA",
                                  "OTHERS": "OTROS", "SANITIZER": "DESINFECTANTE", "LIQUID & GEL": "LÍQUIDO Y GEL", "POWDER": "POLVO", 
                                  "PRETREAT": "PRETRATAMIENTO", "BAR": "BARRA"})
DIM_SEGMENT
DIM_SEGMENT.to_csv('DIM_SEGMENT.csv', index=False)
```

```python
# DataFrame "DIM_CALENDAR"

# Todas las tablas tienen los nombres de sus columnas bien escritas por lo que los errores tipograficos no existen
# En caso de que si, podemos renombrarlas
# Para este DataFrame en cuestión solo cambiaremos los nombres a español
DIM_CALENDAR.columns = ["SEMANA", "AÑO", "MES", "# SEMANA", "FECHA"]

# Verificamos si existen valores nulos con ".isnull().sum()" y de ser así, se rellenaran con "NO VALIDO" 
DIM_CALENDAR.isnull().sum()
DIM_CALENDAR.fillna("NO VALIDO", inplace=True)

# Verificamos si existen datos duplicados
# Eliminar duplicados con la función ".drop_duplicates"
# Al incorporar "inplace=True" aplica los datos directamente en el DataFrame original 
DIM_CALENDAR.drop_duplicates(inplace = True)

DIM_CALENDAR
DIM_CALENDAR.to_csv('DIM_CALENDAR.csv', index=False)
```

```python
# DataFrame "FACT_SALES"

# Todas las tablas tienen los nombres de sus columnas bien escritas por lo que los errores tipograficos no existen
# En caso de que si, podemos renombrarlas
# Para este DataFrame en cuestión solo cambiaremos los nombres a español
FACT_SALES.columns = ["SEMANA", "CODIGO_ARTÍCULO", "VENTAS UNIDADES TOTALES", "VALORES VENTAS TOTALES", "VENTAS SEMANALES PROMEDIO TOTAL POR UNIDAD", "REGION"]

# Verificamos si existen valores nulos con ".isnull().sum()" y de ser así, se rellenaran con "NO VALIDO" 
FACT_SALES.isnull().sum()
FACT_SALES.fillna("NO VALIDO", inplace=True)

# Verificamos si existen datos duplicados
# Eliminar duplicados con la función ".drop_duplicates"
# Al incorporar "inplace=True" aplica los datos directamente en el DataFrame original 
FACT_SALES.drop_duplicates(inplace = True)

# Quitamos los puntos y letras existentes en la columna "CODIGO_ARTÍCULO" para convertirlo a entero
#FACT_SALES['CODIGO_ARTÍCULO'] = FACT_SALES['CODIGO_ARTÍCULO'].str.replace('.', '')
#FACT_SALES['CODIGO_ARTÍCULO'] = FACT_SALES['CODIGO_ARTÍCULO'].str.replace('BP', '')


# Convertir las columnas "CODIGO_ARTÍCULO", "VENTAS UNIDADES TOTALES", "VALORES VENTAS TOTALES", "VENTAS SEMANALES PROMEDIO TOTAL POR UNIDAD" a números
FACT_SALES['CODIGO_ARTÍCULO'] = pd.to_numeric(FACT_SALES['CODIGO_ARTÍCULO'])
FACT_SALES['VENTAS UNIDADES TOTALES'] = pd.to_numeric(FACT_SALES['VENTAS UNIDADES TOTALES'])
FACT_SALES['VALORES VENTAS TOTALES'] = pd.to_numeric(FACT_SALES['VALORES VENTAS TOTALES'])
FACT_SALES['VENTAS SEMANALES PROMEDIO TOTAL POR UNIDAD'] = pd.to_numeric(FACT_SALES['VENTAS SEMANALES PROMEDIO TOTAL POR UNIDAD'])

# No comprendi del todo como estaban organizadas las semanas dentro de la columna "week" (ahora "semanas") así que decidi organizar los datos por 
# cada una de las regiuones en cuestión
# Utilizamos el metodo ".sort_values" para organizalos por region y despues modificamos la tabla original con "inplace=True"
FACT_SALES.sort_values(by = 'REGION',ascending=False, inplace=True)

FACT_SALES.head()
FACT_SALES.to_csv('FACT_SALES.csv', index=False)
```

## Tercera parte 
### Unir los DataFrames relevantes


```python
union1 = pd.merge(
    DIM_PRODUCT,
    DIM_SEGMENT,
    on=['PRESENTACION', "CATEGORIA", "ATRIBUTO_2"],
    how="outer",
    suffixes=('x', 'y')
)

# Paso 2: Complementar los valores "NO DEFINIDO" para ATRIBUTO_1 y ATRIBUTO_3
union1['ATRIBUTO_1'] = union1['ATRIBUTO_1x'].where(union1['ATRIBUTO_1x'] != 'NO DEFINIDO', union1['ATRIBUTO_1y'])
union1['ATRIBUTO_3'] = union1['ATRIBUTO_3x'].where(union1['ATRIBUTO_3x'] != 'NO DEFINIDO', union1['ATRIBUTO_3y'])

# Paso 3: Eliminar columnas auxiliares
resultado = union1.drop(columns=['ATRIBUTO_1x', 'ATRIBUTO_1y', 'ATRIBUTO_3x', 'ATRIBUTO_3y'])

# Mostrar resultado
resultado


# Verifica las claves únicas en ambas tablas
claves_product = DIM_PRODUCT[['PRESENTACION', 'CATEGORIA', 'ATRIBUTO_2']].drop_duplicates()
claves_segment = DIM_SEGMENT[['PRESENTACION', 'CATEGORIA', 'ATRIBUTO_2']].drop_duplicates()

# Encuentra coincidencias
coincidencias = pd.merge(
    claves_product, 
    claves_segment, 
    on=['PRESENTACION', 'CATEGORIA', 'ATRIBUTO_2'], 
    how='inner'
)

print("Total claves únicas en DIM_PRODUCT:", len(claves_product))
print("Total claves únicas en DIM_SEGMENT:", len(claves_segment))
print("Total coincidencias entre tablas:", len(coincidencias))


print(DIM_PRODUCT[['PRESENTACION', 'CATEGORIA', 'ATRIBUTO_2']].info())
print(DIM_SEGMENT[['PRESENTACION', 'CATEGORIA', 'ATRIBUTO_2']].info())


union1 = pd.merge(
    DIM_PRODUCT,
    DIM_SEGMENT,
    on=['PRESENTACION', 'CATEGORIA', 'ATRIBUTO_2'],
    how='outer',
    suffixes=('x', 'y'),
    indicator=True
)

print(union1['_merge'].value_counts())
# Guardar el DataFrame resultado en un archivo .csv
resultado.to_csv('resultado.csv', index=False)
```
