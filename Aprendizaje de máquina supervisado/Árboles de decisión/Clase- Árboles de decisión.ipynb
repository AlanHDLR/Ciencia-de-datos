# Ejemplo 1

```python
fraudes = pd.read_csv("Fraude.csv")
fraudes
```

```python
# Columnas de referencia
feature_cols = ["Reembolso", "EdoCivil", "Ingresos"]
X = fraudes[feature_cols].values
y = fraudes.Fraude
```

```python
# Preprocesamiento para las variables categoricas
from sklearn import preprocessing

Cod_Reembolso = preprocessing.LabelEncoder()
Cod_Reembolso.fit(["Sí", "No"])
X[:,0] = Cod_Reembolso.transform(X[:,0])

Cod_EdoCivil = preprocessing.LabelEncoder()
Cod_EdoCivil.fit(["Soltero", "Casado", "Divorciado"])
X[:,1] = Cod_EdoCivil.transform(X[:,1])
```

```python
X[0:5]
```

```python
# Para los valores de y se requierten variables cualitativas
y[0:5]
```

```python
# Creación de grupos de entrenamiento y prueba
# Cuando las bases de datos son muy grandes, un 20% de prueba basta
# Cuando son muy pequeñas. un 30% es mucho mejor 

# Creación de grupos de prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 1)
```

```python
# Crear objeto de clasificación del árbol de decisión
clf = DecisionTreeClassifier()

#A plicación del algortimo de árboles de decisión a los grupos de entrenamiento
clf = clf.fit(X_train, y_train)

# Predicción de las respuestas para el grupo de prueba
y_pred = clf.predict(X_test)
```

```python
X_test
```

```python
y_pred
```

```python
y_test
```

```python
# Matriz de confusión
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
# Estadisticas de desempeño
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

# Ejemplo 2

```python
pima = pd.read_csv("diabetes.csv")
pima.head()
```

```python
col_names = ["pregnant", "glucose", "bp", "skin", "insulin", "bmi", "pedigree", "age", "label"]
```

```python
pima = pd.read_csv("diabetes.csv", header = None, names = col_names)
pima.head()
```

```python
# Eliminar los ancabezados anteriores
df = pima.iloc[1:,:]
df
```

```python
# Seleccion de caracteristicas
feature_cols = ['pregnant', 'insulin', 'bmi', 'age', 'glucose', 'bp', 'pedigree']
X = df[feature_cols]
y = df.label
```

```python
X
```

```python
y
```

```python
# Creación de grupos de prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
```

```python
# Existen 2 criterios a tener en cuenta
# Criterio de "Gini"
# Criterio de "entriopia"

# Crear objeto de clasificación del árbol de decisión
clf = DecisionTreeClassifier(criterion = 'gini')

#A plicación del algortimo de árboles de decisión a los grupos de entrenamiento
clf = clf.fit(X_train, y_train)

# Predicción de las respuestas para el grupo de prueba
y_pred = clf.predict(X_test)
```

```python
# Matriz de confusión
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
# Estadisticas de desempeño
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
# Creación de información para el árbol (reglas de decisión)
dot_data = tree.export_graphviz(clf, out_file = None, feature_names = feature_cols, class_names = ["Normal", "Diabetes"])

# Creación de la grafica del árbol
graph = pydotplus.graph_from_dot_data(dot_data)
```

# Optimización de desempeño del árbol de decisión

```python
# Se puede definir una maxima profundidad
# Crear objeto de clasificación del árbol de decisión
clf = DecisionTreeClassifier(criterion = 'entropy', max_depth = 3)

#A plicación del algortimo de árboles de decisión a los grupos de entrenamiento
clf = clf.fit(X_train, y_train)

# Predicción de las respuestas para el grupo de prueba
y_pred = clf.predict(X_test)

# Se puede probar con distintos niveles de profundidad ya que no siempre más profundidad = más precisión
```

```python
# Estadisticas de desempeño
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```
