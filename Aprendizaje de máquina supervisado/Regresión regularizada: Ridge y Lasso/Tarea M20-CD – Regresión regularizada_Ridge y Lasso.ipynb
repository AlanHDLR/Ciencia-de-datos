```python
#Primero importamos las librerias que vamos a utilizar y damos la dirección del archivo
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
import warnings
import os
os.chdir("/Users/AlanHDLR/Desktop")
```

```python
# Despues abrimos el archivo y visualizamos los datos 
fuel = pd.read_excel("FuelConsumptionCo2.xlsx", header=None, na_values = ['?'])
fuel.head()
```

```python
# Remover caracteristicas categoricas
# Axis 1 = eje vertical
fuel.drop([0,1,2,3,6,7], axis=1, inplace=True)
# Eliminamos los titulos para que no interfieran en el analisis
fuel.drop([0], axis=0, inplace=True)
fuel
```

```python
# Remover renglones con valores faltantes 
fuel.dropna(inplace = True)

#Convertrimos todos los datos a numericos 
fuel_numerico = fuel.apply(pd.to_numeric, errors='coerce')
print("Tipos de datos después de la conversión robusta:")
print(fuel_numerico.dtypes)

print(fuel)
```

```python
# Separamos la columna de "CO2EMISSIONS" del resto
x = fuel.drop(12, axis=1)
y = fuel[12]
```

```python
# Visualizamos la tabla completa
x
```

```python
# Visualizamos la columna de "C02EMISSIONS"
y
```

```python
# Division en grupos de entrenamiento (training) y de prueba (testing)
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.30, random_state=1)
```

```python
#Construiremos un modelo de regresion lineal multiple
from sklearn.linear_model import LinearRegression
linreg = LinearRegression()
linreg.fit(x_train, y_train)
```

```python
# Impresion de los coeficientes de regresión
print("Intercepto: ", linreg.intercept_)
print("Coeficientes: ", linreg.coef_)
```

```python
# Predicciones
y_pred = linreg.predict(x_test)
y_pred
```

```python
# Calculo de indicadores de la bondad de ajuste
from sklearn.metrics import r2_score
from sklearn import metrics
```

```python
# Impresion de indicadores de bondad de ajuste
# Modelo de regularización
print("Valor de R cuadrada: ", r2_score(y_test, y_pred))
print("Error absoluto medio: ", metrics.mean_absolute_error(y_test, y_pred))
print("Error cuadratico medio: ", metrics.mean_squared_error(y_test, y_pred))
print("Raíz del error cuadratico medio: ", np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
```

# Regresion Ridge

```python
# Alpha = 0 

from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler 
# La libreria Ridge ya no admite la opción "normalize" por lo que tuve que buscar una alternativa, siendo
# esta pipeline 
from sklearn.pipeline import Pipeline           

# Definir el pipeline
# Se utiliza StandardScaler para estandarizar los datos (media 0, desviación estándar 1),
ridgered_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('ridgered', Ridge(alpha=0)) 
])

# Entrenar el pipeline

ridgered_pipeline.fit(x_train, y_train)

# Predecir con el pipeline
y_pred = ridgered_pipeline.predict(x_test)

# Impresion de indicadores de bondad de ajuste
# Modelos de regularización
print("Valor de R cuadrada: ", r2_score(y_test, y_pred))
print("Error absoluto medio: ", metrics.mean_absolute_error(y_test, y_pred))
print("Error cuadratico medio: ", metrics.mean_squared_error(y_test, y_pred))
print("Raíz del error cuadratico medio: ", np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
```

```python
# Examinación de los coeficientes de la regresión Ridge

# Accediendo al modelo "Ridge" dentro del pipeline:
ridge_model = ridgered_pipeline.named_steps['ridgered']

# Imprimimos los atributos del modelo "Ridge":
print("Intercepto: ", ridge_model.intercept_)
print("Coeficientes: ", ridge_model.coef_)
```

```python
# Definición de un rango de prueba de Alpha
alpha_range = 10.**np.arange(-2,3)
alpha_range
```

```python
# Definir el rango optimo de Alpha
from sklearn.linear_model import RidgeCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
import numpy as np

# Definir el rango de alpha:
alpha_range = np.logspace(-4, 0, 50) 

# Crear el pipeline
# El pipeline primero estandariza los datos y luego aplica "RidgeCV"
ridgeregcv_pipeline = Pipeline([
    # El escalador debe ir primero
    ('scaler', StandardScaler()),
    ('ridgeregcv', RidgeCV(alphas=alpha_range, 
                           scoring='neg_mean_squared_error'))
])

# Entrenar el pipeline
ridgeregcv_pipeline.fit(x_train, y_train)

# Acceder al mejor alpha encontrado (ridgeregcv.alpha_)
best_alpha = ridgeregcv_pipeline.named_steps['ridgeregcv'].alpha_

print("Mejor valor de Alpha (encontrado por CV): ", best_alpha)
```

```python
# Prediccion mediante el mejor valor de Alpha
y_pred = ridgeregcv_pipeline.predict(x_test)

# Impresion de indicadores de bondad de ajuste
print("Valor de R cuadrada: ", r2_score(y_test, y_pred))
print("Error absoluto medio: ", metrics.mean_absolute_error(y_test, y_pred))
print("Error cuadratico medio: ", metrics.mean_squared_error(y_test, y_pred))
print("Raíz del error cuadratico medio: ", np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
```

# Regresión Lasso

```python
# Este regresión se encarga de eliminar los coeficientes que no considere aptos (los más pequeños)
# Prueba con un valor de Alpha  = 0.001
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler # 1. Importar el escalador
from sklearn.pipeline import Pipeline           # 2. Importar el Pipeline

# Definir el pipeline
# La estandarización es necesaria para la regresion Lasso, por lo que usamos StandardScaler.
lassoreg_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('lasso', Lasso(alpha=0.001)) 
])

# Entrenar el pipeline
# El pipeline maneja la estandarización de x_train y luego el ajuste del modelo.
lassoreg_pipeline.fit(x_train, y_train)

# Acceder al modelo Lasso 
# Debemos referenciar el pipeline actual ('lassoreg_pipeline').
lasso_model = lassoreg_pipeline.named_steps['lasso']

# Imprimimos el Intercepto y Coeficientes
print("Intercepto: ", lasso_model.intercept_)
print("Coeficientes: ", lasso_model.coef_)
```

```python
# Prueba con un valor de Alpha  = 0.01
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler 
from sklearn.pipeline import Pipeline          

# Definir el pipeline

lassoreg_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('lasso', Lasso(alpha=0.01)) 
])

# Entrenar el pipeline
lassoreg_pipeline.fit(x_train, y_train)

# Acceder al modelo Lasso 
lasso_model = lassoreg_pipeline.named_steps['lasso']

# Imprimimos Intercepto y Coeficientes
print("Intercepto: ", lasso_model.intercept_)
print("Coeficientes: ", lasso_model.coef_)
```

```python
# Determinación del grado de bondad de ajuste con Alpha = 0.01
y_pred = lassoreg_pipeline.predict(x_test)

#Impresion de indicadores de bondad de ajuste
print("Valor de R cuadrada: ", r2_score(y_test, y_pred))
print("Error absoluto medio: ", metrics.mean_absolute_error(y_test, y_pred))
print("Error cuadratico medio: ", metrics.mean_squared_error(y_test, y_pred))
print("Raíz del error cuadratico medio: ", np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
```

```python
# Seleccion del valor optimo de Alpha para la regresión de Lasso
from sklearn.linear_model import LassoCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
import numpy as np # Asegúrate de que numpy esté disponible

# Definir el rango de alphas (debes tener 'alpha_range' definido, por ejemplo):
alpha_range = np.logspace(-4, 0, 50) 

# Paso 1: Crear el Pipeline
# El Pipeline primero estandariza los datos y luego aplica LassoCV
lassoregcv_pipeline = Pipeline([
    # El escalador debe ir primero
    ('scaler', StandardScaler()),
    
    # El modelo LassoCV. Se elimina 'normalize=True'.
    # LassoCV usa 'eps', 'n_alphas' y 'cv' por defecto.
    # No es necesario especificar 'scoring' a menos que cambies el valor predeterminado 
    # de MSE (Mean Squared Error), ya que LassoCV tiene una optimización propia.
    ('lassoregcv', LassoCV(alphas=alpha_range))
])

# Paso 2: Entrenar el Pipeline
lassoregcv_pipeline.fit(x_train, y_train)

# Paso 3: Acceder al mejor alpha encontrado (lassoregcv.alpha_)
# Acceder al modelo anidado a través de named_steps.
best_alpha = lassoregcv_pipeline.named_steps['lassoregcv'].alpha_

print("Mejor valor de Alpha (encontrado por CV): ", best_alpha)
```

```python
# Predicción mediante regresion de Lasso con Alpha optimo
# Determinación del grado de bondad de ajuste con Alpha = 0.01
y_pred = lassoregcv_pipeline.predict(x_test)

#Impresion de indicadores de bondad de ajuste
print("Valor de R cuadrada: ", r2_score(y_test, y_pred))
print("Error absoluto medio: ", metrics.mean_absolute_error(y_test, y_pred))
print("Error cuadratico medio: ", metrics.mean_squared_error(y_test, y_pred))
print("Raíz del error cuadratico medio: ", np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
```
