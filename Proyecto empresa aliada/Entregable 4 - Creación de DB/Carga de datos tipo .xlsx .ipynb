# Importación de datos para DIM_CALENDAR

```python
import pandas as pd
from sqlalchemy import create_engine
import urllib

# Configuración de conexión
server = 'DESKTOP-2U14VQ3' 
database = 'Proyecto_empresa_aliada'
params = urllib.parse.quote_plus(f'DRIVER={{ODBC Driver 17 for SQL Server}};SERVER={server};DATABASE={database};Trusted_Connection=yes;')
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

# Cargar el Excel
df = pd.read_excel('DIM_CALENDAR.xlsx')

# Visualizar los primeros 5 registros 
df.head()
```

```python
# Convertimos a string y quitamos espacios extra
df['WEEK'] = df['WEEK'].astype(str).str.strip()
```

```python
# Hacemos un submit para cargar los datos
try:
    df.to_sql('DIM_CALENDAR', con=engine, if_exists='append', index=False)
    print("Importación exitosa")
except Exception as e:
    print(f"Error: {e}")
```
